# Explicação de Máquinas de Vetores Suporte

O SVM é usado em vários aplicativos, como detecção de faces, detecção de intrusão, classificação de e-mails, artigos de notícias e páginas da Web, classificação de genes e reconhecimento de escrita à mão. Oferece uma precisão muito alta em comparação com outros classificadores, como regressão logística e árvores de decisão. Reconhecido por seu truque de kernel para lidar com espaços de entrada não lineares.

O classificador separa os pontos de dados usando um hiperplano com a maior quantidade de margem. É por isso que um classificador **SVM** também é conhecido como **classificador discriminativo**. O SVM encontra um hiperplano ideal que ajuda a classificar novos pontos de dados.

## Máquinas de Vetor de Suporte
São consideradas uma abordagem de classificação, mas podem ser empregadas em ambos os tipos de problemas de classificação e regressão. Pode lidar facilmente com diversas variáveis contínuas e categóricas. O SVM constrói um hiperplano no espaço multidimensional para separar classes diferentes. O SVM gera um hiperplano ideal de forma iterativa, que é usado para minimizar um erro. A ideia central do SVM é encontrar um hiperplano marginal máximo (MMH) que melhor divida o conjunto de dados em classes.

**Hiperplano:** fronteira de decisão que separa os dados em diferentes classes. A linha (em 2D) ou plano (em 3D) que divide o espaço para classificar os pontos.

## Vetores de suporte
Os vetores de suporte são os pontos de dados que estão mais próximos do hiperplano. Esses pontos definirão melhor a linha de separação por meio do cálculo das margens. Esses pontos são mais relevantes para a construção do classificador.

## Margem
Uma margem é um espaço entre as duas linhas nos pontos de classe mais próximos. Isso é calculado como a distância perpendicular da linha aos vetores de suporte ou aos pontos mais próximos. Se a margem for maior entre as classes, ela será considerada uma boa margem; uma margem menor é uma margem ruim.

# Como funciona?

O principal objetivo é segregar o conjunto de dados fornecido da melhor maneira possível. A distância entre os dois pontos mais próximos é conhecida como margem. O objetivo é selecionar um hiperplano com a margem máxima possível entre os vetores de suporte no conjunto de dados fornecido. O SVM procura o hiperplano marginal máximo nas etapas a seguir:

1. Gerar hiperplanos que segreguem as classes da melhor maneira.

    Na imagem: 3 hiperplanos: preto, azul e laranja.

    O azul e o laranja têm maior erro de classificação, mas o preto está separando as duas classes corretamente.

2. Selecione o hiperplano com a segregação máxima dos dois pontos de dados mais próximos.

Todavia, não é possível lidar com **planos não lineares e inseparáveis** usando hiperplanos lineares.

Nessa situação, o SVM usa um truque de kernel para transformar o espaço de entrada em um espaço de dimensão mais alta, conforme a parte da direita da imagem. Os pontos de dados são plotados nos eixos **x** e **z** (**z** é a soma ao quadrado de **x** e **y**: **z**=**y**^2=**y**^2). Agora pode-se segregar facilmente esses pontos usando a separação linear.

## SVM Kernels (Truque de Kernel)


O algoritmo SVM é implementado na prática usando um kernel, técnica chamada truque de kernel, que transforma um espaço de dados de entrada no formato necessário, ajudando a criar um classificador mais preciso. O kernel pega um espaço de entrada de baixa dimensão e o transforma em um espaço de dimensão mais alta. Em outras palavras, ele converte problemas não separáveis em problemas separáveis, acrescentando mais dimensão a eles. Útil em problemas de separação não linear.
- **Kernel linear:** pode ser usado como produto de ponto normal de duas observações dadas. O produto entre dois vetores é a soma da multiplicação de cada par de valores de entrada.

  `K(x, xi) = sum(x * xi)`

- **Kernel polinominal:** uma forma mais generalizada do kernel linear. O kernel polinomial pode distinguir espaços de entrada curvos ou não lineares.

  `K(x,xi) = 1 + sum(x * xi)^d`

  **d** é o grau do polinômio. **d**=1 é semelhante à transformação linear. O grau precisa ser especificado manualmente no algoritmo de aprendizado.

- **Kernel de função de base radial:** uma função de kernel popular comumente usada na classificação de máquinas de vetor de suporte. A RBF pode mapear um espaço de entrada em um espaço dimensional infinito.

  `K(x,xi) = exp(-gamma * sum((x – xi^2)))`

  **gamma** é um parâmetro, que varia de 0 a 1. Um valor mais alto de **gamma** se ajustará perfeitamente ao conjunto de dados de treinamento, o que causa um ajuste excessivo. **Gamma**=0,1 é considerado um bom valor padrão. O valor de **gamma** precisa ser especificado manualmente no algoritmo de aprendizado.


## Analogia das Frutas (Truqe de Kernel)

🎨 **A Analogia das Frutas na Mesa**

  Imagine que você colocou sobre a mesa de jantar Laranjas 🟠 e Limões 🟢. O seu objetivo é colocar uma régua reta na mesa para separar as laranjas de um lado e os limões do outro.

  - **Cenário 1 (Tudo fácil):** as laranjas estão do lado esquerdo e os limões do lado direito. Você coloca a régua no meio e pronto.

  - **Cenário 2 (O problema):** as laranjas estão todas juntas no centro da mesa, e os limões estão espalhados em volta delas, formando um círculo.

  Não importa como você coloque a régua (linha reta) sobre a mesa, é impossível separar. Se você tentar cortar, vai sempre pegar laranjas e limões juntos.

  É aqui que entram os Kernels (o Truque do Kernel). Eles são como "mágicas" que nos ajudam a ver esse problema de ângulos diferentes.

🛠️ **Os 3 Tipos de Kernels Explicados com Exemplos**
  1. **Kernel Linear (A Régua Simples)**

      É a forma mais básica. Ele não faz nenhuma "mágica" de dimensão extra. Ele simplesmente tenta desenhar uma linha reta (ou um plano) no formato em que os dados já estão.

      Quando usar: Quando seus dados já são fáceis de separar (como no Cenário 1).

      Exemplo: Separar e-mails entre Spam e Não-Spam baseado em poucas palavras-chave bem claras (ex: se tem a palavra "PIX grátis", é spam).

  2. **Kernel Polinomial (A Régua Dobrada / Curva)**

      Voltando ao problema do círculo de limões ao redor das laranjas: como uma linha reta não resolve, o Kernel Polinomial "curva" a nossa visão dos dados. Ele consegue criar fronteiras de decisão que parecem arcos, parábolas ou curvas suavemente dobradas.

      O parâmetro **d (Grau)**:

      **d** = 1: É uma linha reta simples (igual ao Kernel Linear).

      **d** = 2: Permite criar curvas simples (como uma parábola ou uma elipse/círculo para cercar as laranjas).

      **d** = 3 ou mais: Permite criar ondas e curvas cada vez mais complexas.

     Exemplo: Prever se um cliente vai cancelar a assinatura de um serviço cruzando a Idade com a Frequência de Uso. Às vezes, pessoas muito jovens OU muito idosas cancelam mais, gerando um padrão em formato de curva ("U").

  3. **Kernel RBF - Função de Base Radial (A Terceira Dimensão / A Mágica)**

      Este é o Kernel mais famoso e usado porque resolve os problemas mais difíceis.

      Imagine que, em vez de tentar dobrar uma linha na mesa, você joga as laranjas no ar. As laranjas ganham altura/3ª dimensão, enquanto os limões continuam na mesa.
      Agora, você consegue passar uma folha de papel (um hiperplano) no ar, no espaço entre a mesa e as laranjas flutuando. Você separou perfeitamente!

      O parâmetro **gamma (A Sensibilidade)**:

      **gamma** baixo (ex: 0.1): O "jogar" levanta um grupo inteiro de uma vez de forma suave. O modelo fica genérico e calmo (bom para aprender a regra geral).

      **gamma** muito alto (ex: 1.0 ou mais): O "jogar" cria "picos" individuais para cada fruta. O modelo tenta isolar ponto por ponto perfeitamente. Isso gera o **Overfitting** (o modelo "décora" os dados de treino e erra tudo quando chegam frutas novas).

      Exemplo: Reconhecimento facial ou diagnósticos médicos complexos, onde as características estão completamente misturadas e é impossível achar um padrão reto ou uma curva simples.

**Vantagens**
* **Eficaz em Alta Dimensão: mantém um ótimo desemepnho mesmo quando o conjunto de dados possui muyito mais colunas (atributos) do que linhas (registros).
* **Eficiência de Memória:** como só precisa guaradr os **Vetores de Suporte** (aqueles pontos-chave da fronteira) para tomar decisões, o modelo final ocupa pouco espaço na memória.
* **Poder do Truque do Kernel(Kernel Trick):** é extremamente versátil, pois consegue resolver problemas complexos e altamente não-lineares projetando os dados para dimensões superiores com baixo custo computacional.
* **Imunidade a Overfitting na Margem Rígida:** o foco em maximizar a margem de separação ajuda o modelo a generalizar muito bem para dados novos, evitando decorar os dados de treino.
* **Garantia de Solução Global:** diferente de redes neurais, a otimização matemática da SVM encontra um único ponto ótimo (mínimo global), garantindo que o treinamento não fique preso em soluções "meio certas".

**Limitações**
* **Dificl de Interpretar e Visualizar:** ao contrário das Árvores de Decisão, uma SVM (especialmente com funções de Kernel) funciona como uma "caixa preta". É muito difícil explicar matematicamente para um leigo o porquê de uma decisão em espaços multidimensionais.
* **Demorado em Grandes Datasets:** o tempo de treinamento cresce drasticamente conforme o número de registros aumenta, tornando o algoritmo lento para o cenário de Big Data.
* **Sensível a Ruídos e Outliers:** em margem rígida, um único ponto errado ou rebelde na fronteira pode entortar ou arruinar o hiperplano de separação.
* **Complexidade na Escolha do Kernel:** não existe uma regra fixa para saber qual funçãode Kernle funcionará melhor (Linear, RBF, Polinomial); o cientista de dados precisa descobrir por tentativa e erro.
* **Não Fornece Probabilidades Diretamente:** a SVM entrega apenas um veredito seco (ex: "É a Classe 1"). Para obter a porcentagem de certeza (probabilidade), é preciso rodar cálculos extras pesados por fora (como a calibração de Platt).

## Importações: Bibliotecas e Dependências



```python
from sklearn import datasets
from sklearn.model_selection import train_test_split
from sklearn import svm
from sklearn import metrics
```

## Importar Dataset

```python
# Carregar dataset
cancer = datasets.load_breast_cancer()
```

## Explorando os dados

```python
print("Features:", cancer.feature_names)

print("Labels:", cancer.target_names)
```

```python
cancer.data.shape
```

```python
# Registros do conjunto de recursos.
print(cancer.data[0:5])
```

```python
# Conjunto de metas - cancer labels (0:malignant, 1:benign)
print(cancer.target)
```

## Divisão de dados

```python
# Separar dataset entre conjunto de treinamento e teste
x_treinamento, x_teste, y_treinamento, y_teste = train_test_split(cancer.data, cancer.target, test_size=0.3, random_state=109) # 30% teste, 70% treinamento
```

## Modelo de geração

```python
# Criando a SVM de classificação
clf = svm.SVC(kernel='linear')

# Treinando o modelo
clf.fit(x_treinamento, y_treinamento)

# Prevendo a resposta para o conjunto de teste
y_prev = clf.predict(x_teste)
```

## Avaliação do modelo

```python
# Acurácia do modelo: quão frequente o classificador acerta?
print("Acurácia:", metrics.accuracy_score(y_teste, y_prev)*100)
```

```python
# Precisão do modelo
print("Precisão:", metrics.precision_score(y_teste, y_prev)*100)
```

```python
# Recall do modelo
print("Recall:", metrics.recall_score(y_teste, y_prev)*100)
```

```python
import pandas as pd

"""
Exibir os coeficientes para entender a importância das features
Em um SVM linear, os coeficientes indicam a importância de cada feature na decisão.
clf.coef_ é uma array 2D onde as linhas correspondem às classes e as colunas às features.
Para o problema binário (0 e 1), há apenas uma linha de coeficientes.
"""

coefficients = clf.coef_[0]
feature_names = cancer.feature_names

coef_df = pd.DataFrame({'Feature': feature_names, 'Coefficient': coefficients})
coef_df = coef_df.sort_values(by='Coefficient', ascending=False)

print("Coeficientes do Modelo SVM (influência de cada feature):")
print(coef_df)
```
